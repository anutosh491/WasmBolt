## Reuse definitions across cells

Run the next two cells in order to define `square` and call it. Then change
the argument in the call and run that cell again. Restart the kernel to
clear its definitions and start over.


```cpp
int square(int value) { return value * value; }
```

```cpp
std::cout << "square(7): " << square(7) << "\n";
```

```c
#include <stdio.h>

int square(int value) { return value * value; }
```

```c
printf("C square(7): %d\n", square(7));
```

## Reuse definitions across cells

Definitions stay available in later cells until you restart the kernel.
Change the argument in `square(7)` above and run that cell again.
The next cell uses an array and a loop to add three values.


```c
int values[] = {6, 12, 24};
int total = 0;
for (int i = 0; i < 3; ++i) {
  total += values[i];
}
printf("C total: %d\n", total);
```
